## Gis

```python
gpd_villes = gpd.GeoDataFrame(
    villes,
    geometry=gpd.points_from_xy(villes.longitude_centre, villes.latitude_centre),
    crs='EPSG:4326'
)
```

```python
gpd_villes.total_bounds
```

```python
gpd_villes31 = gpd_villes.loc[gpd_villes.dep_code == '31']
gpd_villes31.total_bounds
```

```python
import folium
minx, miny, maxx, maxy = gpd_villes31.total_bounds
centre = [(miny + maxy) / 2, (minx + maxx) / 2]

m = folium.Map(location=centre, zoom_start=9, tiles='CartoDB positron')

folium.GeoJson(
    gpd_villes31,
    tooltip=folium.GeoJsonTooltip(
        fields=['nom_standard', 'population', 'superficie_km2'],
        aliases=['Ville', 'Population', 'Superficie (km²)']
    )
).add_to(m)
m
```

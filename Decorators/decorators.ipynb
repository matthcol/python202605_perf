````python
from functools import wraps

# my custom decorator

def debug(f):
    @wraps(f)
    def decorator(*args, **kwargs):
        print(f"[BEGIN] args={args}, kwargs={kwargs}")
        res = f(*args, **kwargs)
        print(f"[END] result={res}")
        return res
    return decorator
````

````python
@debug
def f(a, b, *c, d=3, e=None, **kwargs):
    "c'est la doc officielle de f"
    return a + b + sum(c) + d
````

````python
y = f(3, 4, activity='piscine apreo petanque')
y
````

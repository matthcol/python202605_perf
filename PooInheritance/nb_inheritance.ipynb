## Héritage et numpy

```python
data = np.random.normal(10.0, 2.5, 1_000_000)
assert isinstance(data, np.ndarray)
assert isinstance(data, object)
data
```

```python
data.min(), data.max()
```

```python
data[data < 0]
```

```python
data_m = np.ma.masked_less(data, 0)
data_m
```

```python
for d in data, data_m:
    stats = {
        'min': d.min(),
        'max': d.max(),
        'mean': d.mean(),
        'std': d.std()
    }
    print(stats)
```

```python
assert type(data_m) is np.ma.MaskedArray
assert type(data_m) is not np.ndarray

assert isinstance(data_m, np.ma.MaskedArray)
assert isinstance(data_m, np.ndarray)
assert isinstance(data_m, object)
```

## Héritage et exception

```python
data = {
    'population': 1000,
    'nom': 'Toulouse'
}
for k, n in [
    ('population', 4), 
    ('nom', 4),
    ('dummy', 4),
    ('population', 0),
]:
    try:
        res = data[k] // n
        print('[INFO]', k, n, res)
    # except Exception as ex:
    #     print('pb', type(ex), ex)
    except KeyError as ex:
        print('[ERROR]', 'info manquante', type(ex), ex)
    except TypeError as ex:
        print('[ERROR]', 'info non divisable', type(ex), ex)
    except ArithmeticError as ex: # inclus ZeroDivisionError
        print('[ERROR]', 'erreur de calcul', type(ex), ex)
```
